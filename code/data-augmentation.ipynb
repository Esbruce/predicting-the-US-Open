### Instantiating data columns for useful extra statistics

#### Looping through the data

```python
k = 32
default_elo = 1500

# sort chronologically (use whatever your date column is actually called)
df = df.sort_values('tourney_date')

# drop rows with missing player IDs, make IDs strings, fill missing surfaces
df = df.dropna(subset=['winner_id', 'loser_id'])
df['winner_id'] = df['winner_id'].astype(str)
df['loser_id'] = df['loser_id'].astype(str)
df['surface'] = df['surface'].fillna('Unknown')

# reset the index LAST so it's a clean 0..n-1 for df.at[match_idx, ...]
df = df.reset_index(drop=True)

elo = {}        # player_id -> current elo
h2h = {}        # (player_a, player_b) -> list of past winners
surf_h2h = {}   # (player_a, player_b, surface) -> list of past winners

winner_elo, loser_elo = [], []
winner_h2h, loser_h2h = [], []
winner_surf_h2h, loser_surf_h2h = [], []


def calc_h2h_rate(history, player):
    last_5 = history[-5:]          # last 5 meetings only
    if len(last_5) == 0:
        return 0.5                 # no history: neutral
    return last_5.count(player) / len(last_5)


# loop through each match in the data
for match_idx in range(len(df)):
    winner = df.at[match_idx, 'winner_id']
    loser = df.at[match_idx, 'loser_id']
    surface = df.at[match_idx, 'surface']

    pair = tuple(sorted((winner, loser)))
    surf_pair = pair + (surface,)

    # --- read features from BEFORE this match ---
    w_elo = elo.get(winner, default_elo)
    l_elo = elo.get(loser, default_elo)
    winner_elo.append(w_elo)
    loser_elo.append(l_elo)

    w_h2h = calc_h2h_rate(h2h.get(pair, []), winner)
    winner_h2h.append(w_h2h)
    loser_h2h.append(1 - w_h2h)

    w_surf = calc_h2h_rate(surf_h2h.get(surf_pair, []), winner)
    winner_surf_h2h.append(w_surf)
    loser_surf_h2h.append(1 - w_surf)

    # --- now update state with this match's result ---
    expected_winner = 1 / (1 + 10 ** ((l_elo - w_elo) / 400))
    elo[winner] = w_elo + k * (1 - expected_winner)
    elo[loser] = l_elo - k * (1 - expected_winner)

    h2h[pair] = h2h.get(pair, []) + [winner]
    surf_h2h[surf_pair] = surf_h2h.get(surf_pair, []) + [winner]


df['winner_elo'] = winner_elo
df['loser_elo'] = loser_elo
df['winner_h2h_last_5'] = winner_h2h
df['loser_h2h_last_5'] = loser_h2h
df['winner_surface_h2h_last_5'] = winner_surf_h2h
df['loser_surface_h2h_last_5'] = loser_surf_h2h

df.to_csv('FeatureAugmentedData.csv', index=False)
```

### Randomise Player Data to be symetrical. Winner expected to be player A 50 % of time. Thus non-bias

```python
def write_to_new_data(df, path="PreppedData.csv", seed=42):
    groups = [
        ("winner_", "loser_", ["id", "seed", "entry", "hand", "ht", "ioc", "age",
                               "rank", "rank_points", "elo",
                               "h2h_last_5", "surface_h2h_last_5"]),
        ("w_", "l_", ["ace", "df", "svpt", "1stIn", "1stWon", "2ndWon",
                      "SvGms", "bpSaved", "bpFaced"]),
    ]

    rng = np.random.default_rng(seed)
    winner_is_A = rng.random(len(df)) < 0.5

    data = df[["tourney_id", "tourney_date", "surface", "tourney_level",
               "indoor", "best_of", "round", "draw_size"]].copy()

    for w_prefix, l_prefix, features in groups:
        for f in features:
            w, l = df[f"{w_prefix}{f}"], df[f"{l_prefix}{f}"]
            data[f"A_{f}"] = np.where(winner_is_A, w, l)
            data[f"B_{f}"] = np.where(winner_is_A, l, w)

    for f in ["age", "ht", "rank", "rank_points", "elo",
              "h2h_last_5", "surface_h2h_last_5"]:
        data[f"diff_{f}"] = data[f"A_{f}"] - data[f"B_{f}"]

    data["label"] = winner_is_A.astype(int)  # 1 = A won

    data.to_csv(path, index=False)
    return data
```
